```python
data_combined=[]

for file in files:

        data=pd.read_csv(folder+'/'+file)
        data.dropna(inplace=True, ignore_index=True) 
        data['Start accelero file']=START
        data['file_name']=file
        data['Observation date']=file.split('-')[3]+' '+file.split('-')[4] #get Observation date from the name of the file rather because a lot of mistake from Boris
        data.rename(columns={"Started": "Start (s)", "Ended": "Stop (s)"},inplace=True)
        
        data['Start (s)']=pd.to_datetime(data['Start (s)'], format='%H:%M:%S:%f')
        data['Start (s)']=data['Start (s)'].dt.hour*3600+data['Start (s)'].dt.minute*60 + data['Start (s)'].dt.second + 0.000001*data['Start (s)'].dt.microsecond
        data['Stop (s)']=pd.to_datetime(data['Stop (s)'], format='%H:%M:%S:%f')
        data['Stop (s)']=data['Stop (s)'].dt.hour*3600+data['Stop (s)'].dt.minute*60 + data['Stop (s)'].dt.second + 0.000001*data['Stop (s)'].dt.microsecond
        data['Duration (s)']=   data['Stop (s)']-   data['Start (s)']
        
        data=data[['Start accelero file','file_name','Observation date','Start (s)', 'Stop (s)', 'Duration (s)','Behavior']]
        
        data['datetime_deb']=pd.to_datetime(data['Observation date'], format='%Y%m%d %H%M%S')+pd.to_timedelta(data['Start (s)'], unit='s')
        data['datetime_fin']=pd.to_datetime(data['Observation date'], format='%Y%m%d %H%M%S')+pd.to_timedelta(data['Stop (s)'], unit='s')
        data['Start (s)']=[(data['datetime_deb'][x]-START).total_seconds() for x in range(data.shape[0])]
        data['Stop (s)']=[(data['datetime_fin'][x]-START).total_seconds() for x in range(data.shape[0])]
        data['Observation date']=pd.to_datetime(data['Observation date'], format='%Y%m%d %H%M%S')

        # Separate date and time
        data["date_deb"] = data["datetime_deb"].dt.strftime("%Y-%m-%d")
        data["time_deb"] = data["datetime_deb"].dt.strftime("%H:%M:%S.%f").str[:-3]
        data["time_fin"] = data["datetime_fin"].dt.strftime("%H:%M:%S.%f").str[:-3]
        
   
        
    
        data_combined.append(data)
```

```python
Behaviors_rblt=pd.DataFrame(np.concatenate(data_combined, axis=0))
Behaviors_rblt.columns=data.columns
```

```python

Behaviors_rblt.to_csv('C:/Users/loren/Documents/Tortues/Martinique_Verte/Behavior_rblt/'+file_name+'_rblt.csv',sep=';',index=False)
```
